```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict,Annotated
from langchain_core.messages import HumanMessage,BaseMessage
from langchain_google_genai import ChatGoogleGenerativeAI
from langgraph.graph.message import add_messages
from dotenv import load_dotenv
from langgraph.prebuilt import ToolNode,tools_condition
from langchain_community.tools import DuckDuckGoSearchRun
from langchain_core.tools import tool
import requests
import random
import os
```

```python
load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

```python
search_tool=DuckDuckGoSearchRun(region="us-en")
```

```python
@tool
def calculator(first_number: float, second_number: float, operation: str) -> dict:
    """Perform basic arithmetic operations.

    Supported operations:
    - add
    - sub
    - mul
    - div
    """
    try:
        if operation == "add":
            result = first_number + second_number
        elif operation == "sub":
            result = first_number - second_number
        elif operation == "mul":
            result = first_number * second_number
        elif operation == "div":
            if second_number == 0:
                return {"error": "Division by zero is not allowed."}
            result = first_number / second_number
        else:
            return {"error": f"Unsupported operation: {operation}"}

        return {
            "first_number": first_number,
            "second_number": second_number,
            "operation": operation,
            "result": result,
        }

    except Exception as e:
        return {"error": str(e)}
```

```python
@tool
def get_stock_price(symbol: str) -> dict:
    """Get the latest stock price for a given stock symbol."""

    api_key = os.getenv("ALPHA_VANTAGE_API_KEY")

    url = (
        "https://www.alphavantage.co/query"
        f"?function=GLOBAL_QUOTE"
        f"&symbol={symbol}"
        f"&apikey={api_key}"
    )

    try:
        response = requests.get(url)
        response.raise_for_status()
        return response.json()

    except Exception as e:
        return {"error": str(e)}
```

```python
tools=[get_stock_price,calculator,search_tool]
llm_with_tools=llm.bind_tools(tools)
tool_node=ToolNode(tools)
```

```python
class ChatState(TypedDict):
    messages:Annotated[list[BaseMessage],add_messages] #Annotated[type,metadata]
    
```

```python
def chat_node(state: ChatState):
    messages = state["messages"]
    response = llm_with_tools.invoke(messages)
    return {"messages": [response]}
```

```python
graph=StateGraph(ChatState)
graph.add_node("chat_node",chat_node)
graph.add_node("tools",tool_node)
```

```python
graph.add_edge(START,"chat_node")
graph.add_conditional_edges("chat_node",tools_condition)
graph.add_edge("tools","chat_node")
```

```python
chatbot=graph.compile()
chatbot
```

```python
output=chatbot.invoke({"messages":[HumanMessage(content="hello!")]})
```
