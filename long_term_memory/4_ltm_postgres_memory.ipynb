```python
from langgraph.store.postgres import PostgresStore
```

```python
import os
from dotenv import load_dotenv

load_dotenv()

DB_URI = os.getenv(
    "DATABASE_URL",
    "postgresql://postgres:postgres@localhost:5432/langgraph_db"
)
```

```python
store = PostgresStore.from_conn_string(DB_URI)
```
