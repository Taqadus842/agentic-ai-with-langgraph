```python
from langgraph.store.memory import InMemoryStore
```

```python
store=InMemoryStore()
```

```python
namespace=("user","u1")
```

Creating Memory

```python
store.put(namespace,"1",{"data":"User likes pizza"})
```

```python
store.put(namespace,"2",{"data":"User prefers dark mode"})
```

```python
namespace2=("user","u2")
store.put(namespace2,"1",{"data":"User like pasta"})
store.put(namespace2,"2",{"data":"User prefers grid style navigation"})
```

Retrieving Memories

```python
store.get(namespace,"1")
```

Retrieving all Memories

```python
items=store.search(namespace)
for item in items:
    print(item.value)
```

```python
items=store.search(namespace2)
for item in items:
    print(item.value)
```

Semantic Search

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from dotenv import load_dotenv

load_dotenv()

embeddingmodel = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001"
)
```

```python
store = InMemoryStore(
    index={
        "embed": embeddingmodel,
        "dims": 768
    }
)
```

```python
namespace=('users':'u1')
```

```python
store.put(namespace, "1", {"data": "User likes pizza"})
store.put(namespace, "2", {"data": "User prefers Python for programming"})
store.put(namespace, "3", {"data": "User enjoys working on AI projects"})
store.put(namespace, "4", {"data": "User is interested in Generative AI"})
store.put(namespace, "5", {"data": "User likes reading research papers"})
store.put(namespace, "6", {"data": "User prefers practical programming examples"})
store.put(namespace, "7", {"data": "User is learning LangGraph"})
store.put(namespace, "8", {"data": "User works with RAG systems"})
store.put(namespace, "9", {"data": "User is interested in AI agents"})
store.put(namespace, "10", {"data": "User prefers simple explanations"})
```

```python
items=store.search(namespace,query="What is user currently leanring?",limit=1)
for item in items:
    print(item.value)
```
